```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_excel("data/grouped_31_lines_mean_and_signed_error.xlsx", header=0, sheet_name="Data & Signed Errors")

data.head()
```

Compute the canonical coordinates

Compute covariance matrix

```python
errors = data[["value_2_signed_error", "value_3_signed_error"]]

covariance = errors.cov()

print(covariance)
```

```python
graph = data.plot.scatter(x = "value_2_signed_error", y = "value_3_signed_error", title="Graph of X and Y Errors (AprilTag)")

graph.axhline(0, color="black", linestyle="--", linewidth=1)
graph.axvline(0, color="black", linestyle="--", linewidth=1)

plt.show()
```
